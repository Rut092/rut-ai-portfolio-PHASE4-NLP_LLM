## 1. Comparing Pre-trained Hugging Face Tokenizers (BERT vs. GPT-2)

```python
# hugging face has different model tokenizors that contains lets use it
from transformers import AutoTokenizer
```

## Using Already Available Tokenizer

```python
bert_tok = AutoTokenizer.from_pretrained("bert-base-cased")
gpt2_tok = AutoTokenizer.from_pretrained("gpt2")
```

```python
text = "I've been waiting for a HuggingFace course my whole life. the unhappiest dachshund defenstrated anthropomorphically"
```

```python
print(bert_tok.tokenize(text)) # this is bert token
print(gpt2_tok.tokenize(text)) # this is gpt2
```

```python
# so bert users wordpiece encoding
# and gpt2 uses BPE encoding

encoded = bert_tok(text)
print("Token IDs")
print(encoded.input_ids)
print()
print('Decoded back')
print(bert_tok.decode(bert_tok.encode(text)))
```

```python
sentences = [
    "I've been waiting for a HuggingFace course my whole life.",
    "I hate this so much!",
    "😂 added some emojis too",
    "Hello 💡 World !",
    "don't cant won't",
    "café résumé naïve",
    "int main() { return 0; }"
]

for sent in sentences:
    tokens = bert_tok.tokenize(sent)
    print(f'{sent}\n')
    print(f'Tokens:({len(tokens)}) : {tokens}\n')
    print('---------------------------------------------------')
```

```python
from tokenizers.models import BPE # Imports the BPE model for byte-pair encoding tokenization.
from tokenizers.trainers import BpeTrainer # Imports the trainer for the BPE model, used to train a new tokenizer.
from tokenizers.pre_tokenizers import Whitespace # Imports a pre-tokenizer that splits text by whitespace.
from tokenizers import Tokenizer # Imports the base Tokenizer class, used to build custom tokenizers.

# Difference: 'tokenizers.models.BPE' defines the BPE algorithm, 'tokenizers.trainers.BpeTrainer' trains a tokenizer using this algorithm, 'tokenizers.pre_tokenizers.Whitespace' is a step before tokenization, and 'tokenizers.Tokenizer' is the overarching class to combine these components.
```

## 2. Building and Training a Custom Hindi BPE Tokenizer

```python
demo_text_file = """भारत एक विविध और सांस्कृतिक रूप से समृद्ध देश है। यहाँ कई भाषाएँ बोली जाती हैं, जिनमें हिंदी सबसे व्यापक रूप से समझी और बोली जाने वाली भाषा है।
भारतीय संस्कृति कला, संगीत, नृत्य और व्यंजनों का एक अद्भुत मिश्रण है जो सदियों पुराने इतिहास से प्रेरित है।
आज के डिजिटल युग में, हिंदी भाषा का उपयोग इंटरनेट और विभिन्न सोशल मीडिया प्लेटफॉर्म पर बहुत तेजी से बढ़ रहा है।
प्राकृतिक भाषा प्रसंस्करण (NLP) और मशीन लर्निंग तकनीकों का उपयोग करके हम हिंदी पाठ का विश्लेषण और टोकनाइजेशन आसानी से कर सकते हैं।
यह तकनीक हिंदी साहित्य, समाचार और ऑनलाइन सामग्री को बेहतर ढंग से समझने और उसका अनुवाद करने में मदद करती है।
शिक्षा, विज्ञान और प्रौद्योगिकी के क्षेत्रों में भी हिंदी का महत्व निरंतर बढ़ रहा है।"""

# Save this text to a file so that we can train the tokenizer on it
with open("hindi_corpus.txt", "w", encoding="utf-8") as f:
    f.write(demo_text_file)

print("Hindi demo text dataset successfully loaded and saved to 'hindi_corpus.txt'.")
```

### Loading and Preparing the Hindi Corpus

```python
tokenizer = Tokenizer(BPE(unk_token="[UNK]"))
tokenizer.pre_tokenizer = Whitespace()

trainer = BpeTrainer(
    special_tokens=["[UNK]", "[CLS]", "[SEP]", "[PAD]", "[MASK]"],
    min_frequency=2,
    vocab_size=1000,
)

tokenizer.train(files = ["hindi_corpus.txt"],trainer = trainer)

# Encoding a sample Hindi sentence from our corpus
sample_sentence = "भारत एक समृद्ध देश है और हिंदी बहुत व्यापक भाषा है।"
output = tokenizer.encode(sample_sentence)
print('Original Text:', sample_sentence)
print('Tokens:', output.tokens)
print('IDs:', output.ids)
print('Decoded:', tokenizer.decode(output.ids))
```

### Saving and Loading the Custom Tokenizer

```python
tokenizer.train(files=["hindi_corpus.txt"], trainer=trainer)
tokenizer.save("hindi_tokenizer.json")
# never training again its stored in a json file
# 2. Just load the pre-built dictionary
from tokenizers import Tokenizer
ready_tokenizer = Tokenizer.from_file("hindi_tokenizer.json")

# 3. Now you CAN encode directly in microseconds:
output = ready_tokenizer.encode("भारत एक स्वतंत्र देश है।")
print(output.ids)
```

## Inspecting the Saved Tokenizer's Vocab and Merges

```python
import json

# 1. Load the tokenizer json file as a standard Python dictionary
with open("hindi_tokenizer.json", "r", encoding="utf-8") as f:
    tokenizer_data = json.load(f)

# 2. Extract key components
# The vocabulary (text-to-ID mapping) and merges (rules for fusing characters)
vocab = tokenizer_data["model"]["vocab"]
merges = tokenizer_data["model"]["merges"]

# 3. Print the top elements to see what the BPE algorithm learned!
print("--- FIRST 15 VOCAB ENTRIES ---")
# Let's show the first 15 items in our vocab dictionary
for i, (token, token_id) in enumerate(list(vocab.items())[:50]):
    print(f"Row {token_id:03d}: {token}")

print("\n--- FIRST 10 MERGE RULES ---")
# Merges show which tokens get fused together. High-priority merges are at the top.
for i, merge_rule in enumerate(merges[:50]):
    print(f"Rule {i+1:02d}: {merge_rule}")
```
